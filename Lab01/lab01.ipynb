# Ex 1. b

```python
import matplotlib.pyplot as plt

fig, axs = plt.subplots(3)
fig.suptitle('Titlu')

x = np.cos(520 * np.pi * t + np.pi / 3)
axs[0].plot(t, x)
y = np.cos(280 * np.pi * t - np.pi / 3)
axs[1].plot(t, y)
z = np.cos(120 * np.pi * t + np.pi / 3)
axs[2].plot(t, z)

for ax in axs.flat:
    ax.set_xlim([0, 0.03])
```

# Ex 1. c.

```python
T = 1 / 200
N = int(0.03 * 200) + 1
n = np.linspace(0, N - 1, N)
tn = n * T

fig, axs = plt.subplots(3)
fig.suptitle('Titlu')

xn = np.cos(520 * np.pi * tn + np.pi / 3)
axs[0].stem(tn, xn)
yn = np.cos(280 * np.pi * tn - np.pi / 3)
axs[1].stem(tn, yn)
zn = np.cos(120 * np.pi * tn + np.pi / 3)
axs[2].stem(tn, zn)

for ax in axs.flat:
    ax.set_xlim([0, 0.03])
```

> Dacă un eșanțion este memorat pe 4 biți, câți, i bytes vor ocupa 1 oră de achiziție?

sample_rate * 1h = 2000 * 60m = 2000 * 3600s = 7,200,000 (7 milioane 200 de mii) de sampleuri

7.2mil * 4 bits = 28.8mil bits
28.8mil bits / 8 = 3.6mil bytes = 3.6MB

```python
print(1 / 2000)

sample_n = 2000 * 3600
print(sample_n)

bytes_n = sample_n * 4 / 8
print(bytes_n)
```
